## 1. Design Envelope in Subsea Asset Integrity

### Definition

A **design envelope** is the range of operating and environmental conditions within which a subsea asset is designed to operate safely and reliably throughout its intended service life.

The design envelope considers parameters such as:

* Pressure and temperature.
* Flow rate and fluid composition.
* External hydrostatic pressure.
* Environmental loads, including waves and currents.
* Mechanical stresses and fatigue loads.
* Corrosion allowance and material strength.
* Installation, operation, shutdown, and restart conditions.

### Why the P-T Envelope Is Critical

The **pressure-temperature (P-T) envelope** defines the allowable combinations of pressure and temperature for a production system.

It is particularly important because exceeding the envelope can result in:

1. **Overpressure:** Internal pressure exceeding the design limit can cause yielding, leakage, or pipeline rupture.
2. **Temperature-related material limitations:** High temperatures can reduce material strength and affect seals and other components.
3. **Hydrate formation:** Low temperatures combined with suitable pressure and fluid composition can cause gas hydrates to form and potentially block the flowline.
4. **Wax formation:** Cooling may cause wax deposition, restricting flow.
5. **Thermal stress:** Temperature changes can produce expansion, contraction, and additional mechanical stress.
6. **Integrity degradation:** Repeated pressure and temperature cycles can contribute to fatigue and accelerate some degradation mechanisms.

## 2. Minimum Wall Thickness Using Barlow's Formula

### Given Data

* Design pressure, \(P = 4,467\text{ psi}\)
* Outside diameter, \(D = 10.75\text{ inches}\)
* Pipe grade: API 5L X65
* Specified minimum yield strength, \(SMYS = 65,000\text{ psi}\)

### Formula

A simplified form of Barlow's formula is:

$$
t = \frac{PD}{2SF}
$$

Where:

* \(t\) = required wall thickness, in inches
* \(P\) = internal pressure, in psi
* \(D\) = outside diameter, in inches
* \(S\) = allowable stress, in psi
* \(F) = designed factor, float = 0.72

For an initial screening calculation, assume the allowable stress equals the specified minimum yield strength:

$$
S = 65,000\text{ psi}
$$

### Calculation

Substitute the given values:

$$
t = \frac{4,467 \times 10.75}{2 \times 65,000 \times 0.72}
$$

Calculate the numerator:

$$
4,467 \times 10.75 = 48,020.25
$$

Calculate the denominator:

$$
2 \times 65,000 \times 0.72 = 93,000
$$

Therefore:

$$
t = \frac{48,020.25}{93,000}
$$

$$
t = 0.516\text{ inches}
$$

Convert inches to millimetres:

$$
t = 0.513 \times 25.4
$$

$$
\boxed{t \approx 13.03\text{ mm}}
$$

### Engineering Interpretation

The simplified Barlow calculation gives a nominal pressure-related thickness estimate of approximately **9.38 mm**.

However, this is not automatically the final code-compliant minimum wall thickness. Actual pipeline design must consider the applicable design code, design factor, material factor, weld-joint factor where applicable, manufacturing tolerances, corrosion allowance, external pressure, and other relevant load cases.

For a preliminary design calculation, the result is useful as a screening estimate. The final design thickness must be established using the governing pipeline design standard and its specified allowable-stress methodology.

## 3. Hydrate Subcooling and Risk Classification

### Definition of Hydrate Subcooling

**Hydrate subcooling** is the difference between the hydrate equilibrium temperature and the actual operating temperature at a given pressure and fluid composition.

It indicates how far the operating temperature lies below the hydrate formation boundary.

The formula is:

$$
\Delta T_{\text{sub}} = T_{\text{eq}} - T_{\text{op}}
$$

Where:

* \(\Delta T_{\text{sub}}\) = hydrate subcooling, in °C
* \(T_{\text{eq}}\) = hydrate equilibrium temperature at the operating pressure and fluid composition, in °C
* \(T_{\text{op}}\) = actual operating temperature, in °C

### Given Data

* Operating pressure: \(P = 3,500\text{ psi}\)
* Operating temperature: \(T_{\text{op}} = 15^\circ\text{C}\)

### Calculating the Equilibrium Temperature

The question refers to a simplified Hammerschmidt model but does not provide the model equation, coefficients, or the gas composition.

A hydrate equilibrium temperature cannot be uniquely calculated from pressure and operating temperature alone. It also depends on fluid composition and the selected thermodynamic model.

Therefore, the equilibrium temperature must be obtained from the specific simplified model provided in the course material or from an appropriate hydrate-equilibrium calculation.

Suppose the prescribed simplified model gives:

$$
T_{\text{eq}} = 20^\circ\text{C}
$$

This value is an assumed example, not a verified result for the Usan field or the stated operating conditions.

### Calculate Subcooling



$$
\Delta T_{\text{sub}} = T_{\text{eq}} - T_{\text{op}}
$$

$$
\Delta T_{\text{sub}} = 20^\circ\text{C} - 15^\circ\text{C}
$$

$$
\boxed{\Delta T_{\text{sub}} = 5^\circ\text{C}}
$$

### Risk Classification

The actual risk classification depends on the thresholds specified by the course or operating company.

For the illustrative equilibrium temperature of \(20^\circ\text{C}\), the operating temperature of \(15^\circ\text{C}\) lies below the hydrate equilibrium temperature. Hydrate formation is therefore thermodynamically possible if the required gas and water phases are present.

A subcooling of \(5^\circ\text{C}\) indicates a potential hydrate formation risk. However, it is not possible to assign a definitive low, medium, or high classification without the prescribed risk matrix and additional information about fluid composition, water availability, inhibitors, and operating conditions.

## 4. Four Primary Corrosion Mechanisms in Subsea Carbon Steel Pipelines

Corrosion is the deterioration of metal caused by chemical or electrochemical interactions with its environment.

Four important corrosion mechanisms or forms relevant to subsea carbon steel pipelines are described below.

| Corrosion mechanism | Description                                                                           | Driving physical or chemical parameter                                                      | Corresponding monitoring sensor or technique                                                         |
| ------------------- | ------------------------------------------------------------------------------------- | ------------------------------------------------------------------------------------------- | ---------------------------------------------------------------------------------------------------- |
| General corrosion   | Relatively widespread metal loss across a surface                                     | Corrosion rate, influenced by temperature, pH, corrosive species, and water chemistry       | Corrosion coupons, electrical-resistance probes, and ultrasonic thickness measurements               |
| Pitting corrosion   | Localised corrosion producing pits or cavities                                        | Local electrochemical conditions, chloride concentration, and breakdown of protective films | High-resolution ultrasonic inspection, specialised inspection tools, and localised corrosion probes  |
| CO₂ corrosion       | Corrosion caused by dissolved carbon dioxide in produced water                        | CO₂ partial pressure, water chemistry, pH, temperature, and flow conditions                 | Corrosion probes, water-chemistry analysers, pH sensors, and pressure-temperature instrumentation    |
| Erosion-corrosion   | Metal loss caused by the combined effects of flowing fluids, particles, and corrosion | Fluid velocity, turbulence, sand concentration, and corrosive fluid chemistry               | Sand-erosion monitoring systems, flowmeters, corrosion probes, and ultrasonic thickness measurements |

### Explanation

**1. General corrosion**

General corrosion produces relatively uniform metal loss over an exposed surface. Over time, it reduces the remaining wall thickness and can weaken the pipeline.

**2. Pitting corrosion**

Pitting corrosion produces localised cavities. Although the total metal loss may be small, an individual pit can penetrate the pipe wall and initiate leakage.

**3. CO₂ corrosion**

Carbon dioxide dissolves in produced water and can form carbonic acid, contributing to corrosion of carbon steel. The rate depends on several variables, including water chemistry, temperature, pressure, and flow conditions.

**4. Erosion-corrosion**

Erosion-corrosion results from the combined effects of mechanical material removal and electrochemical corrosion. High fluid velocities, turbulence, and entrained sand can increase metal loss in susceptible locations.

### Application to Predictive AI

An AI model can combine sensor measurements and inspection records to estimate corrosion rates, identify high-risk locations, and predict when wall thickness may approach its minimum allowable value.

The model should distinguish between general metal loss and localised damage because average wall-thickness measurements may not adequately capture the risk from pitting.

## 5. Corrosion Rate, Remaining Life, and Inspection Interval

Assuming a constant average corrosion rate:

$$
CR = \frac{t_0-t_c}{\text{Exposure time}}
$$

Substitute the values:

$$
CR = \frac{12.7-10.8}{10}
$$

$$
CR = \frac{1.9}{10}
$$

Therefore:

$$
\boxed{CR = 0.19\text{ mm/year}}
$$

### B. Calculate Remaining Life

The remaining wall thickness available before reaching the code-minimum thickness is:

$$
t_{\text{remaining}} = t_c-t_{\min}
$$

Substitute:

$$
t_{\text{remaining}} = 10.8-8.5
$$

$$
\boxed{t_{\text{remaining}} = 2.3\text{ mm}}

Estimated remaining life is:

$$
RL = \frac{t_c-t_{\min}}{CR}

$$

$$
RL = \frac{2.3}{0.19}
$$

$$
\boxed{RL \approx 12.1\text{ years}}

* Corrosion-rate uncertainty.
* Measurement accuracy and inspection coverage.
* The possibility of localised pitting.

* Consequences of failure.
* Applicable integrity-management requirements.

For example, if the measured corrosion rate increases, the inspection interval should be shortened.

## 6. FMECA and Its Role in Predictive AI

### Definition of FMECA

**Failure Modes, Effects, and Criticality Analysis (FMECA)** is a systematic engineering method used to identify possible failure modes, examine their effects, and assess their criticality.

It helps engineers understand how an asset can fail, what causes each failure, what the consequences might be, and which failures require the greatest attention.

For subsea assets, FMECA can be applied to pipelines, risers, jumpers, pumps, valves, compressors, and other equipment.

### Purpose of FMECA in Predictive AI

FMECA provides an engineering foundation for designing predictive AI models.

Its main purposes include:

1. Identifying the failure modes that the model should predict.
2. Identifying physical and operational parameters associated with each failure.
3. Prioritising failures according to likelihood and consequences.
4. Selecting relevant sensors and historical inspection data.
5. Defining appropriate prediction targets.
6. Helping engineers interpret model predictions.
7. Supporting maintenance planning and risk reduction.

### Failure Mode-to-Feature Mapping

A machine-learning feature is an input variable used by a model to make a prediction.

Failure mode-to-feature mapping connects each engineering failure mechanism to the variables that may reveal its development.

For example:

| Failure mode       | Potential contributing factors                     | Candidate ML features                                                      | Prediction target                                                |
| ------------------ | -------------------------------------------------- | -------------------------------------------------------------------------- | ---------------------------------------------------------------- |
| Internal corrosion | Water chemistry, CO₂, temperature, pressure        | Water cut, pH, CO₂ partial pressure, temperature, corrosion-probe readings | Corrosion rate or probability of wall-thickness limit exceedance |
| Hydrate blockage   | High pressure, low temperature, water availability | Pressure, temperature, flow rate, water content, inhibitor concentration   | Hydrate formation risk                                           |
| Fatigue cracking   | Cyclic loading and stress concentration            | Stress range, pressure cycles, vibration, cycle count                      | Fatigue damage or crack-growth risk                              |
| Erosion-corrosion  | High velocity and entrained particles              | Flow velocity, sand concentration, differential pressure                   | Erosion rate or remaining wall thickness                         |

### How Mapping Informs ML Model Design

The mapping determines which inputs should be collected, how the target variable should be defined, and which model is appropriate.

For example, if the objective is to predict whether a pipeline will experience a critical wall-thickness reduction within 30 days, the training data should include historical sensor readings and inspection information aligned with that prediction horizon.

The model should not be trained using variables that would only become available after the failure occurs.

FMECA also helps prevent a common modelling mistake: using the same features for every failure mode even when different physical mechanisms require different information.

# Fatigue Life Estimation of a D-Class Weld Using the DNV S-N Curve

## Problem Statement

A D-class weld in seawater with cathodic protection experiences a stress range of 60 MPa. Calculate the predicted number of cycles to failure using the DNV S-N curve.

## Given Data

| Parameter                 | Symbol           |                             Value |
| ------------------------- | ---------------- | --------------------------------: |
| Weld classification       | -                |                           D-class |
| Environment               | -                | Seawater with cathodic protection |
| Stress range              | $\Delta \sigma$  |                            60 MPa |
| First-region slope        | $m_1$            |                                 3 |
| Second-region slope       | $m_2$            |                                 5 |
| First-region coefficient  | $\log_{10}(a_1)$ |                            11.764 |
| Second-region coefficient | $\log_{10}(a_2)$ |                            15.606 |

## S-N Curve Equation

The general S-N curve equation is:

$$
\log_{10}(N) = \log_{10}(a) - m\log_{10}(\Delta \sigma)
$$

Where:

* $N$ = number of cycles to fatigue failure.
* $\Delta \sigma$ = applied stress range in MPa.
* $m$ = slope of the S-N curve.
* $a$ = S-N curve coefficient.

For this calculation, the assumed D-class curve coefficients are:

**Region 1:**

$$
\log_{10}(N) = 11.764 - 3\log_{10}(\Delta \sigma)
$$

**Region 2:**

$$
\log_{10}(N) = 15.606 - 5\log_{10}(\Delta \sigma)
$$

## Calculation

### Step 1: Evaluate the First Region

Substitute the stress range:

$$
\log_{10}(N) = 11.764 - 3\log_{10}(60)
$$

Since:

$$
\log_{10}(60) \approx 1.77815
$$

Then:

$$
\log_{10}(N) = 11.764 - (3 \times 1.77815)
$$

$$
\log_{10}(N) \approx 6.42955
$$

Therefore:

$$
N = 10^{6.42955} \approx 2.69 \times 10^6 \text{ cycles}
$$

This exceeds $10^6$ cycles, so the first-region result is outside its assumed range. Evaluate the second region.

### Step 2: Evaluate the Second Region

Use the second-region equation:

$$
\log_{10}(N) = 15.606 - 5\log_{10}(60)
$$

Substitute the logarithm:

$$
\log_{10}(N) = 15.606 - (5 \times 1.77815)
$$

$$
\log_{10}(N) = 15.606 - 8.89075
$$

$$
\log_{10}(N) \approx 6.71525
$$

Convert to cycles:

$$
N = 10^{6.71525}
$$

$$
\boxed{N \approx 5.19 \times 10^6 \text{ cycles}}
$$

## Final Result

The predicted fatigue life is approximately:

$$
\boxed{5.19 \text{ million cycles}}
$$

This is an estimate based on the stated S-N curve coefficients and assumptions.

## 6. Engineering Considerations

Actual fatigue life may be affected by:

* Weld geometry and stress concentration.
* Thickness correction factors.
* Variable-amplitude loading.
* Fabrication quality and weld defects.
* The applicable DNV standard edition and curve definition.

**Note:** Verify the curve coefficients, stress units, and applicable S-N curve region against the official DNV standard before using this result in an engineering design or integrity assessment.

### Engineering Interpretation

The illustrative calculation gives approximately 4.63 million cycles to failure.

**This is not a verified DNV D-class fatigue-life prediction.** The assumed coefficient is included only to demonstrate the calculation method. The actual result must be calculated using the specified DNV S-N curve for the relevant weld class and environment.

Cathodic protection does not automatically mean that a weld follows a particular curve without further qualification. The applicable environmental curve and any required corrections must be established from the governing standard.

## 8. Miner's Rule and Estimated Fatigue Life

### Definition of Miner's Rule

Miner's rule is a linear cumulative fatigue-damage model. It estimates accumulated fatigue damage by summing the damage contributions from different stress ranges.

The damage ratio for a stress range is:

$$
D_i=\frac{n_i}{N_i}
$$

Where:

* \(D_i\) = damage ratio for stress range \(i\)
* \(n_i\) = number of cycles actually experienced at that stress range
* \(N_i\) = number of cycles to failure at that stress range

Total cumulative damage is:

$$
D_{\text{total}}=\sum_i D_i
$$

Failure is conventionally associated with:

$$
D_{\text{total}}\approx1
$$

This is a simplified engineering assumption; actual fatigue failure can occur at damage levels different from one.

### Given Data

A riser has accumulated the following damage ratios over five years:

| Stress range | Accumulated damage ratio |
| ------------ | -----------------------: |
| 30 MPa       |                     0.15 |
| 50 MPa       |                     0.08 |
| 80 MPa       |                     0.03 |

### A. Calculate Total Damage

Using Miner's rule:

$$
D_{\text{total}}=0.15+0.08+0.03
$$

$$
\boxed{D_{\text{total}}=0.26}
$$

The accumulated damage is 0.26, equivalent to 26% of the conventional linear cumulative-damage failure threshold.

### B. Estimate Total Fatigue Life

Assume:

1. The calculated damage represents the damage accumulated over five years.
2. Future loading remains statistically similar to the historical loading.
3. The damage accumulation rate remains constant.

The average damage accumulation rate is:

$$
\dot D=\frac{D_{\text{total}}}{t}
$$

Substitute:

$$
\dot D=\frac{0.26}{5}
$$

$$
\boxed{\dot D=0.052\text{ per year}}
$$

The estimated time to reach a cumulative damage ratio of one is:

$$
T_{\text{total}}=\frac{1}{\dot D}
$$

Substitute:

$$
T_{\text{total}}=\frac{1}{0.052}
$$

$$
\boxed{T_{\text{total}}\approx19.23\text{ years}}
$$

This is the estimated total fatigue life from the start of the assumed loading history.

## 9. Why Physics-Informed ML Models Can Outperform Purely Data-Driven Models

### Definition

A purely data-driven machine-learning model learns patterns from historical data without necessarily incorporating the governing physical laws of the engineering system.

A **physics-informed ML model** combines data-driven learning with engineering principles, physical constraints, or mathematical relationships.

Examples of these principles include pressure-vessel equations, corrosion-rate relationships, heat-transfer equations, fluid mechanics, and fatigue-damage accumulation.

### Advantages of Physics-Informed ML

**1. Better performance when data are limited**

Subsea assets may have relatively few confirmed failures because equipment is designed for long service lives and failures are comparatively uncommon.

Physics-based relationships can provide useful constraints when the available failure dataset is small.

**2. Improved physical consistency**

A model can be designed to respect known physical limits. For example, it should not predict a negative remaining wall thickness.

**3. Better interpretability**

Engineers can compare predictions against established physical relationships and investigate discrepancies.

**4. Improved generalisation**

When operating conditions change, a model that incorporates physical knowledge may generalise better than a model that relies entirely on correlations in historical data. This benefit is not guaranteed and must be validated.

**5. Better maintenance decisions**

Physics-based predictions can help engineers understand not only whether an asset is at risk but also which degradation mechanism is responsible.

### Example 1: Pipeline Corrosion Prediction

A purely data-driven model could use historical temperature, pressure, and inspection records to predict future wall thickness.

A physics-informed model could additionally use the measured corrosion rate and a wall-thickness degradation relationship:

$$
t_{\text{future}}=t_{\text{current}}-CR\times\Delta t
$$

Where:

* \(t_{\text{future}}\) = estimated future wall thickness
* \(t_{\text{current}}\) = current measured wall thickness
* \(CR\) = estimated corrosion rate
* \(\Delta t\) = elapsed time

The model can predict when the wall thickness may approach the minimum allowable value.

For example, if the current thickness is 10.8 mm, the minimum allowable thickness is 8.5 mm, and the corrosion rate is 0.19 mm/year, the simplified remaining-life estimate is:

$$
RL=\frac{10.8-8.5}{0.19}
$$

$$
RL\approx12.1\text{ years}
$$

An ML model can refine this estimate using changing operating conditions and new inspection measurements.

### Example 2: Riser Fatigue Prediction

A purely data-driven model might learn relationships between vibration measurements, stress ranges, and observed fatigue damage.

A physics-informed model could use S-N curves and Miner's rule to estimate cumulative fatigue damage:

$$
D_{\text{total}}=\sum_i\frac{n_i}{N_i}
$$

The model could combine measured stress cycles with calculated fatigue resistance to predict damage accumulation and estimate when inspection or maintenance may be required.

## 10. Importance of Data Dictionaries in Engineering ML Pipelines

### Definition

A **data dictionary** is a structured document that defines the meaning, format, units, valid values, and other important properties of variables used in a dataset.

In an engineering ML pipeline, it helps data scientists and engineers understand what each sensor measurement represents and how it should be interpreted.

### Why Data Dictionaries Are Essential

**1. Consistent interpretation**

A variable such as `pressure` could represent internal pressure, external hydrostatic pressure, or differential pressure. A data dictionary removes this ambiguity.

**2. Correct units**

Pressure may be recorded in psi, bar, or MPa. A documented unit prevents accidental comparisons or calculations using incompatible units.

**3. Data-quality control**

The dictionary can define valid ranges and identify implausible measurements, missing values, or sensor errors.

**4. Integration of multiple data sources**

Different systems may use different names for the same physical quantity. A data dictionary supports consistent mapping across databases and equipment-monitoring systems.

**5. Reproducibility**

Documented variables allow different engineers and data scientists to reproduce the same transformations and understand how model inputs were prepared.

**6. Model reliability**

Incorrect variable interpretation can lead to misleading model predictions. A data dictionary helps prevent these errors.

### Five Required Fields in an Engineering Data Dictionary

A useful engineering data dictionary should contain at least the following five fields for every variable.

| Field                         | Description                                     | Example                                                           |
| ----------------------------- | ----------------------------------------------- | ----------------------------------------------------------------- |
| 1. Variable name              | Unique identifier used in the dataset or code   | `flowline_pressure`                                               |
| 2. Description or definition  | Explanation of what the variable measures       | Internal pressure of a subsea production flowline                 |
| 3. Unit of measurement        | Standard unit in which the variable is recorded | psi                                                               |
| 4. Data type                  | Format of the stored value                      | Float                                                             |
| 5. Valid range or constraints | Acceptable values or engineering limits         | Non-negative values within the sensor's validated operating range |

### Additional Recommended Fields

For a production-quality engineering ML pipeline, the dictionary should also include:

* Sensor or data source.
* Sampling frequency.
* Timestamp and time zone.
* Missing-value conventions.
* Calibration status.
* Measurement accuracy and uncertainty.
* Data-quality flags.
* Asset identifier and sensor location.
* Preprocessing requirements.
* Whether the variable is an input feature, target variable, or derived feature.
